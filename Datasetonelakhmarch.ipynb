```python
import csv
import random
import string

# S-Box for PRESENT cipher
SBOX = [0xC, 0x5, 0x6, 0xB, 0x9, 0x0, 0xA, 0xD, 0x3, 0xE, 0xF, 0x8, 0x4, 0x7, 0x1, 0x2]

def key_schedule_80(key, round_num):
    round_key = key >> 16
    key = ((key << 61) & (2**80 - 1)) | (key >> 19)
    key = (SBOX[key >> 76] << 76) | (key & (2**76 - 1))
    key ^= round_num << 15
    return round_key, key

def present_encrypt_80(plaintext, key):
    state = plaintext
    for round_num in range(1, 32):
        round_key, key = key_schedule_80(key, round_num)
        state ^= round_key
    round_key, _ = key_schedule_80(key, 32)
    return state ^ round_key

def generate_printable_string(length):
    return ''.join(random.choices(string.printable[10:94], k=length))

num_records = 100000
records = []
plain_records = []
plainkey_records = []
plaincipher_records = []
keys_records = []
cipherkey_records = []
cipherplainkey_records = []
cipherplain_records = []

for i in range(num_records):
    plaintext = generate_printable_string(8)
    if i % 10 == 0:
        key_text = generate_printable_string(10)
        key = int.from_bytes(key_text.encode(), 'big') % (2**80)

    plaintext_int = int.from_bytes(plaintext.encode(), 'big')
    ciphertext = present_encrypt_80(plaintext_int, key)
    ciphertext_chars = ''.join(chr((ciphertext >> (8 * i)) & 0x7F) for i in range(8))
    decrypted_plaintext = plaintext  # Assuming perfect decryption

    records.append([plaintext, key_text, ciphertext_chars, decrypted_plaintext])
    plain_records.append([plaintext])
    plainkey_records.append([plaintext, key_text])
    plaincipher_records.append([plaintext, ciphertext_chars])
    keys_records.append([key_text])
    cipherkey_records.append([ciphertext_chars, key_text])
    cipherplainkey_records.append([ciphertext_chars, plaintext, key_text])
    cipherplain_records.append([ciphertext_chars, plaintext])

def save_csv(filename, header, data):
    with open(filename, "w", newline="") as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(header)
        writer.writerows(data)

save_csv("Encryption.csv", ["Plaintext", "Key", "Ciphertext", "Decrypted Plaintext"], records)
save_csv("plain.csv", ["Plaintext"], plain_records)
save_csv("plainkey.csv", ["Plaintext", "Key"], plainkey_records)
save_csv("plaincipher.csv", ["Plaintext", "Ciphertext"], plaincipher_records)
save_csv("keys.csv", ["Key"], keys_records)
save_csv("cipherkey.csv", ["Ciphertext", "Key"], cipherkey_records)
save_csv("cipherplainkey.csv", ["Ciphertext", "Plaintext", "Key"], cipherplainkey_records)
save_csv("cipherplain.csv", ["Ciphertext", "Plaintext"], cipherplain_records)

print("CSV files generated successfully.")
```
